Create a new notebook and code the following:





1. Create a numpy array of size 2 x 2.

```python
arr1 = np.array(([5,7], [0,9]))
print(arr1)
```

2. Create a random numpy array of size 3 x 4

```python
rand_arr = np.random.rand(3,4)
print(rand_arr)
```

3. Create a numpy array of ones and a numpy array of zeroes

```python
ones = np.ones((10,5))
zeros = np.zeros((6,6))

print(ones, "\n")
print(zeros)
```

4. Create a random numpy array of size 6x3 and transpose it

```python
rand_arr = np.random.rand(6,3)

rand_arr_transpose = rand_arr.transpose()

print(rand_arr_transpose)
```

5. Create two random numpy array of size n x p and p x m and then compute the dot product

```python
n, p, m = 3, 4, 5

rand_arr_1 = np.random.rand(n, p)
rand_arr_2 = np.random.rand(p, m)

dot_product = np.dot(rand_arr_1, rand_arr_2)

print(dot_product)
```
