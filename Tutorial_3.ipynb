```python
#Counts the top 5 nationalities
top_5 = df["nationality"].value_counts().head(5)

plt.bar(top_5.index, top_5.values)

#Add titles, x-axis, y-axis name
plt.title("Top 5 Driver Nationalities")
plt.xlabel("Nationality")
plt.ylabel("Number of Drivers")

#Show the bar chart of Top 5 Driver Nationalities
plt.show()
```
